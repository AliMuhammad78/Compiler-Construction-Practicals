## Program to Generate Quadruples for Any Expression

```python
# Program to Generate Quadruples for Any Expression


expression = "a = b + m * d - o /s + p * q -rw"

quadruples = []
temp_count = 1

# Operator precedence
precedence = {
    '+': 1,
    '-': 1,
    '*': 2,
    '/': 2
}

# Convert expression into tokens
tokens = expression.replace("=", " = ") \
                   .replace("+", " + ") \
                   .replace("-", " - ") \
                   .replace("*", " * ") \
                   .replace("/", " / ").split()

# Left and right side
left_side = tokens[0]
right_side = tokens[2:]

# Infix to Postfix Conversion
output = []
stack = []

for token in right_side:

    if token.isalnum():
        output.append(token)

    elif token in precedence:

        while (stack and stack[-1] in precedence and
               precedence[stack[-1]] >= precedence[token]):
            output.append(stack.pop())

        stack.append(token)

while stack:
    output.append(stack.pop())

# Generate Quadruples
eval_stack = []

for token in output:

    if token.isalnum():
        eval_stack.append(token)

    else:
        arg2 = eval_stack.pop()
        arg1 = eval_stack.pop()

        temp = "t" + str(temp_count)
        temp_count += 1

        quadruples.append([token, arg1, arg2, temp])

        eval_stack.append(temp)

# Final Assignment
final_temp = eval_stack.pop()
quadruples.append(["=", final_temp, "-", left_side])

# Print Quadruples
print("Expression:", expression)

print("\nOp\tArg1\tArg2\tResult")

for q in quadruples:
    print(q[0], "\t", q[1], "\t", q[2], "\t", q[3])
```

# Question No . 18

```python
 

grammar = {
    "S": ["Sb", "b"],
    "A": ["aA", "d"]
}

 
# Program
 

left_recursive = False
ambiguous = False

print("Grammar:\n")

for nt, productions in grammar.items():
    print(nt, "->", " | ".join(productions))

print("\nChecking Grammar...\n")


 
# Check Left Recursion
 

for nt, productions in grammar.items():

    for prod in productions:

        if prod.startswith(nt):

            print(nt, "is Left Recursive")
            left_recursive = True


 
# Check Ambiguity  
# If two productions start with same symbol
 

for nt, productions in grammar.items():

    first_symbols = []

    for prod in productions:

        if len(prod) > 0:
            first_symbols.append(prod[0])

    if len(first_symbols) != len(set(first_symbols)):

        print(nt, "may be Ambiguous")
        ambiguous = True


# ==================================================
# Final Result
# ==================================================

print()

if not left_recursive and not ambiguous:

    print("Grammar is LL(1)")

else:

    print("Grammar is NOT LL(1)")
```

## First and Follow sets


```python


grammar = {
    "S": ["ABC"],
    "A": ["a", ""],
    "B": ["b", ""],
    "C": ["c"]
}

start_symbol = "S"

start_symbol = "S"
# FIRST and FOLLOW Program


FIRST = {}
FOLLOW = {}

# PRINT GRAMMAR


print("Grammar:\n")

for nt, productions in grammar.items():
    print(nt, "->", " | ".join(productions))

 # FIRST SETS


print("\nFIRST Sets:\n")

for nt, productions in grammar.items():

    FIRST[nt] = set()

    for prod in productions:

        first_symbol = prod[0]

        # If first symbol is terminal
        if first_symbol not in grammar:
            FIRST[nt].add(first_symbol)

        # If first symbol is non-terminal
        else:
            FIRST[nt].update(FIRST.get(first_symbol, []))

    print("FIRST(", nt, ") =", list(FIRST[nt]))


 # FOLLOW SETS
 
print("\nFOLLOW Sets:\n")

# Initialize empty sets
for nt in grammar:
    FOLLOW[nt] = set()

# Start symbol gets $
start_symbol = list(grammar.keys())[0]
FOLLOW[start_symbol].add("$")

# Find FOLLOW
for nt, productions in grammar.items():

    for prod in productions:

        for i in range(len(prod)):

            symbol = prod[i]

            # If symbol is Non-Terminal
            if symbol in grammar:

                # If next symbol exists
                if i + 1 < len(prod):

                    next_symbol = prod[i + 1]

                    # Next symbol is terminal
                    if next_symbol not in grammar:
                        FOLLOW[symbol].add(next_symbol)

                    # Next symbol is non-terminal
                    else:
                        FOLLOW[symbol].update(FIRST[next_symbol])

                # If symbol is last
                else:
                    FOLLOW[symbol].update(FOLLOW[nt])

# Print FOLLOW sets
for nt in grammar:
    print("FOLLOW(", nt, ") =", list(FOLLOW[nt]))
```

# Question No. 20

```python


grammar = {
    "S": ["Sa", "Sjdie", "abc", "abd"],
    "A": ["Ac", "yueye", "Awae"]
}

print("Original Grammar:\n")

for nt, productions in grammar.items():
    print(nt, "->", " | ".join(productions))


# Remove Left Recursion

new_grammar = {}

for nt, productions in grammar.items():

    alpha = []
    beta = []

    for prod in productions:

        if prod.startswith(nt):
            alpha.append(prod[len(nt):])

        else:
            beta.append(prod)

    # If left recursion exists
    if alpha:

        new_nt = nt + "'"

        new_grammar[nt] = []
        new_grammar[new_nt] = []

        for b in beta:
            new_grammar[nt].append(b + new_nt)

        for a in alpha:
            new_grammar[new_nt].append(a + new_nt)

        new_grammar[new_nt].append("e")

    else:
        new_grammar[nt] = productions


print("\nAfter Removing Left Recursion:\n")

for nt, productions in new_grammar.items():
    print(nt, "->", " | ".join(productions))


# Left Factoring

factored_grammar = {}

for nt, productions in new_grammar.items():

    prefix_map = {}

    for prod in productions:

        prefix = prod[0]

        if prefix not in prefix_map:
            prefix_map[prefix] = []

        prefix_map[prefix].append(prod)

    changed = False

    for prefix, group in prefix_map.items():

        if len(group) > 1:

            changed = True

            new_nt = nt + "F"

            factored_grammar[nt] = [prefix + new_nt]

            factored_grammar[new_nt] = []

            for item in group:
                factored_grammar[new_nt].append(item[1:])

    if not changed:
        factored_grammar[nt] = productions


print("\nAfter Left Factoring:\n")

for nt, productions in factored_grammar.items():
    print(nt, "->", " | ".join(productions))
```
